### 1. O que é Matplotlib?

O Matplotlib é uma biblioteca Python para criação de visualizações estáticas, animadas e interativas. É uma das ferramentas mais populares para visualização de dados!

```python
# Importando a biblioteca
import matplotlib.pyplot as plt
print("Biblioteca matplotlib.pyplot importada com sucesso!")
```

Por que visualizar dados?

- Identificar padrões rapidamente

- Comunicar informações de forma clara

- Detectar anomalias ou outliers

- Tomar melhores decisões baseadas em dados

### 2. Nosso Primeiro Gráfico: Gráfico de Linha
Vamos começar com algo simples - um gráfico de linha!

```python
import matplotlib.pyplot as plt
import numpy as np

# Dados simples
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Criando o gráfico
plt.plot(x, y)

# Mostrando o gráfico
plt.show()
```

### 3. Customizando Nosso Gráfico

Um gráfico sem informações não é muito útil. Vamos adicionar alguns elementos:



```python
# Dados
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Criando gráfico com customizações
plt.plot(x, y, color='red', linewidth=2, marker='o', markersize=8, label='Crescimento')

# Adicionando informações
plt.title('Meu Primeiro Gráfico Customizado')
plt.xlabel('Eixo X - Tempo')
plt.ylabel('Eixo Y - Valores')
plt.legend()

# Grade para melhor leitura
plt.grid(True, linestyle='--', alpha=0.7)

plt.show()
```

*Parâmetros Úteis*:

- color: cor da linha ('red', 'blue', '#FF5733')

- linewidth: espessura da linha

- marker: estilo do ponto ('o', 's', '^', 'D')

- linestyle: estilo da linha ('-', '--', '-.', ':')

### 4. Gráfico de Barras

Perfeito para comparar categorias!

```python
# Dados de exemplo: Vendas por mês
meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai']
vendas = [150, 200, 180, 300, 250]

plt.bar(meses, vendas, color=['#FF9999', "#7CAAD8", '#99FF99', '#FFD700', '#FF69B4'])
plt.title('Vendas Mensais')
plt.xlabel('Meses')
plt.ylabel('Vendas (unidades)')

# Adicionando valores em cima das barras
for i, v in enumerate(vendas):
    plt.text(i, v + 5, str(v), ha='center')

plt.show()
```

### 5. Gráfico de Dispersão (Scatter Plot)

Ideal para ver relações entre duas variáveis!

```python
# Gerando dados aleatórios
np.random.seed(767554)  # Para resultados consistentes
x = np.random.rand(50) * 100
y = x * 2 + np.random.rand(50) * 30

plt.scatter(x, y, alpha=0.6, c='green', s=50)
plt.title('Relação entre X e Y')
plt.xlabel('Variável X')
plt.ylabel('Variável Y')
plt.grid(True, alpha=0.3)
plt.show()

# Desafio: O que acontece se você mudar o valor de np.random.seed()?
print("Quando você muda o valor de np.random.seed(), o Matplotlib vai gerar um gráfico com pontos totalmente novos e posições diferentes, mas que ainda manterá o mesmo comportamento visual geral (uma linha subindo).")
```

### 6. Múltiplos Gráficos (Subplots)

Às vezes queremos comparar vários gráficos lado a lado:

```python
# Criando figura com subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

# Gráfico 1 - Linha
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

ax1.plot(x, y1, 'b-', label='sen(x)')
ax1.plot(x, y2, 'r--', label='cos(x)')
ax1.set_title('Funções Trigonométricas')
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.legend()
ax1.grid(True)

# Gráfico 2 - Barras
categorias = ['A', 'B', 'C', 'D']
valores = [23, 45, 56, 78]
cores = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4']

ax2.bar(categorias, valores, color=cores)
ax2.set_title('Dados por Categoria')
ax2.set_xlabel('Categorias')
ax2.set_ylabel('Valores')

# Ajustando layout
plt.tight_layout()
plt.show()
```

### 7. Salvando Seu Gráfico

Não esqueça de salvar seu trabalho!

```python
# Criando um gráfico para salvar
x = [1, 2, 3, 4, 5]
y = [x**2 for x in x]  # Quadrado de x

plt.plot(x, y, 'go-', linewidth=2, markersize=8)
plt.title('Gráfico de x²')
plt.xlabel('x')
plt.ylabel('x²')

# Salvando em diferentes formatos
plt.savefig('meu_grafico.png', dpi=300, bbox_inches='tight')
plt.savefig('meu_grafico.pdf')

plt.show()
```

### 8. Atividade Prática

Desafio Final: Crie uma visualização completa usando os conceitos aprendidos!

```python
import matplotlib.pyplot as plt

# Dados para o desafio
disciplinas = ['Matemática', 'Português', 'Ciências', 'História', 'Geografia']
notas_turma_a = [7.5, 8.0, 6.5, 9.0, 7.0]
notas_turma_b = [8.5, 7.0, 8.0, 6.5, 8.5]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

# Gráfico 1 - Barras (Turma A)
ax1.bar(disciplinas, notas_turma_a, color=["#4E54A7", '#66B2FF', "#03CAFC", "#4869FF", "#02F7FFC0"])
ax1.set_title('Disciplinas e suas notas - Turma A')
ax1.set_xlabel('Disciplinas') # Corrigido de 'Notas' para 'Disciplinas'
ax1.set_ylabel('Notas')       # Corrigido de 'Disciplina' para 'Notas'

# Alterado de plt.text para ax1.text
for i, v in enumerate(notas_turma_a):
    ax1.text(i, v + 0.1, str(v), ha='center', va='bottom') 

# Gráfico 2 - Barras (Turma B)
ax2.bar(disciplinas, notas_turma_b, color=["#61DD8A", "#3CFF00", "#0F790F", "#41D13C", "#00FF0DC0"])
ax2.set_title('Disciplinas e suas notas - Turma B')
ax2.set_xlabel('Disciplinas') # Corrigido de 'Notas' para 'Disciplinas'
ax2.set_ylabel('Notas')       # Corrigido de 'Disciplina' para 'Notas'

# Alterado de plt.text para ax2.text
for i, b in enumerate(notas_turma_b):
    ax2.text(i, b + 0.1, str(b), ha='center', va='bottom') 

# Ajustando layout antes de salvar
plt.tight_layout()

# Salvando em diferentes formatos
plt.savefig('comparacao_turmas.png', dpi=300, bbox_inches='tight')
plt.savefig('comparacao_turmas.pdf')

plt.show()
```
